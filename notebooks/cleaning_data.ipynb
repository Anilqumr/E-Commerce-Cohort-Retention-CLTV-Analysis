```python
from pathlib import Path
import pandas as pd
def load_dataset(filename="Ecommerce_Dataset.csv"):
   
    # Project root (assuming notebook is in the notebooks folder)
    project_root = Path.cwd().parent
    file_path = project_root / "data" / "raw" / filename

    if not file_path.exists():
        raise FileNotFoundError(
            f"Dataset not found at:\n{file_path}\n"
            "Ensure the CSV file exists in the data/raw folder."
        )

    Ecommerce_data = pd.read_csv(file_path)
    print(f"Dataset loaded successfully!")
    print(f"Shape: {Ecommerce_data.shape}")

    return Ecommerce_data
```

```python
Ecommerce_data = load_dataset()
Ecommerce_data.head()
```

```python
#Handling missing values
Ecommerce_data.isnull().sum()
print(f"Missing values in each column: {Ecommerce_data.isnull().sum()}")
```

```python
#Checking duplicates
Ecommerce_data.duplicated().sum()
print(f"Number of duplicate rows: {Ecommerce_data.duplicated().sum()}")
```

```python
#Convert to datetime
date_cols = ["transaction_date", "signup_date"]

Ecommerce_data[date_cols] = Ecommerce_data[date_cols].apply(pd.to_datetime)
```

Feature engineering: Creating new features based on existing data can provide additional insights.

```python
#Feature 1: Transaction Month
Ecommerce_data["TransactionMonth"] = (
    Ecommerce_data["transaction_date"]
      .dt.to_period("M")
)
```

```python
#Feature 2: Transaction Year
Ecommerce_data["Year"] = Ecommerce_data["transaction_date"].dt.year
```

```python
#Feature 3: Transaction Month Number
Ecommerce_data["Month"] = Ecommerce_data["transaction_date"].dt.month
```

```python
#Feature 4: Day of Week
Ecommerce_data["DayOfWeek"] = Ecommerce_data["transaction_date"].dt.day_name()
```

```python
#Feature 5: Quarter
Ecommerce_data["Quarter"] = Ecommerce_data["transaction_date"].dt.quarter
```
