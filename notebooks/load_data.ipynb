```python
from pathlib import Path
import pandas as pd
def load_dataset(filename="Ecommerce_Dataset.csv"):
   
    # Project root (assuming notebook is in the notebooks folder)
    project_root = Path.cwd().parent
    file_path = project_root / "data" / "raw" / filename

    if not file_path.exists():
        raise FileNotFoundError(
            f"Dataset not found at:\n{file_path}\n"
            "Ensure the CSV file exists in the data/raw folder."
        )

    Ecommerce_data = pd.read_csv(file_path)
    print(f"Dataset loaded successfully!")
    print(f"Shape: {Ecommerce_data.shape}")

    return Ecommerce_data
```

```python
Ecommerce_data = load_dataset()
Ecommerce_data.head()
```

```python
def explore_dataset(df, name):
   
    print("=" * 60)
    print(f"Dataset: {name}")
    print("=" * 60)

    # Shape
    print(f"\nShape: {df.shape[0]} rows × {df.shape[1]} columns")

    # Column names
    print("\nColumns:")
    print(df.columns.tolist())

    # Data types and missing values
    print("\nDataset Information:")
    print(df.info())

    # Missing values
    print("\nMissing Values:")
    print(df.isnull().sum())

    # Summary statistics
    print("\nSummary Statistics:")
    print(df.describe(include='all').T)

    # First five rows
    print("\nFirst Five Rows:")
    display(df.head())
```

```python
explore_dataset(Ecommerce_data, "Ecommerce Data")
```
