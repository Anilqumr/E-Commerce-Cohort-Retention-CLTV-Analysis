```python
##Notice two things:
#The entire header is inside quotes "...".
#Every row is also inside quotes.
#So Pandas is treating each line as one string instead of splitting it into columns.

df = pd.read_csv(
    "../Data/raw/Ecommerce_Dataset.csv.xls.csv",
    header=None,
    encoding="utf-8-sig"
)
```

```python
##Now assign the correct column names:
df = df[0].str.replace('"', '', regex=False).str.split(",", expand=True)
```

```python
##Now assign the correct column names:
df.columns = [
    "order_id",
    "customer_id",
    "transaction_date",
    "product_category",
    "quantity",
    "unit_price",
    "revenue",
    "signup_date",
    "acquisition_channel",
    "country"
]
```

```python
##Since the original header is now the first row of data, remove it:
df = df.iloc[1:].reset_index(drop=True)
```

```python
##Convert Date Columns
df["transaction_date"] = pd.to_datetime(df["transaction_date"])
df["signup_date"] = pd.to_datetime(df["signup_date"])

##After conversion, Python understands it as a date, allowing us to calculate:

Month
Year
Day
Cohort Month
Retention Period
```

```python
##Convert Quantity
df["quantity"] = df["quantity"].astype(int)

##Why?

Quantity is a whole number.

Example:

4 Products
2 Products
10 Products

Python should store these as integers.
```

```python
##Convert Unit Price
df["unit_price"] = df["unit_price"].astype(float)

##Why?

Prices have decimal values.

Example:

144.98
79.50
22.35

These should be floating-point numbers.
```

```python
##Convert Revenue
df["revenue"] = df["revenue"].astype(float)

#Revenue also contains decimal values.
```

```python
##After that, we'll create the most important column of the project:
## Create Cohort Month

##A cohort is a group of customers who first purchased in the same month.

##We'll create it with:

df["cohort_month"] = (
    df.groupby("customer_id")["transaction_date"]
      .transform("min")
      .dt.to_period("M")
)
```
