```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 1. Load SQL Query Output
df = pd.read_csv('W:\Anil\Infotact Solutions\Project2 Files\Cloning\SaaS-E-Commerce-Cohort-Retention-CLTV-Analysis\Data\Cleaned\sql_cohort_output.csv')
```

```python
# 2. Pivot to build the Cohort Matrix
cohort_matrix = df.pivot_table(
    index='Cohort_Month', 
    columns='cohort_index', 
    values='active_users'
)
```

```python
# 3. Convert to Retention Percentages
cohort_size = cohort_matrix.iloc[:, 0]
retention_matrix = cohort_matrix.divide(cohort_size, axis=0) * 100
```

```python
# 4. Plot Retention Heatmap (Week 4 Deliverable)
plt.figure(figsize=(12, 8))
sns.heatmap(retention_matrix, annot=True, fmt='.1f', cmap='YlGnBu')
plt.title('Cohort Retention Rate (%)')
plt.show()
```
